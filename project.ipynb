## The problem

How does a calculator find $\sqrt{a}$? It makes a guess and improves it again and again. Here are two methods. Stop when the answer is close enough (the error is smaller than 0.000001).

**Newton's method.** Start with `x = a`. Repeat:

$$x_{\text{new}} = \frac{1}{2}\left(x + \frac{a}{x}\right)$$

Stop when `abs(x * x - a) < 0.000001`.

**Example** for `a = 2`: 2 -> 1.5 -> 1.41667 -> 1.414216 -> 1.41421356

**Bisection.** Start with `low = 0` and `high = a` (if `a < 1`, use `high = 1`). Repeat:

- `mid = (low + high) / 2`
- if `mid * mid < a`, then `low = mid`
- otherwise `high = mid`

Stop when `high - low < 0.000001`. The answer is `mid`.

You will count how many steps each method needs.

```python
#The function convert : that we will use to convert when the user inter a string that we can convert into string
def convert(a):  
    if type(a)==int or type(a)==float:
        return a
    elif type(a)!=str:
        return "This is not a number!"
    else:
        a=a.strip()
        sign=1
        if a.startswith("-"):
            sign=-1
            a=a[1:]
        if a.isdigit():
            return sign*int(a)
        if a.count(".")==1:
            parts=a.split(".")
            intpart=parts[0]
            decipart=parts[1]
            if intpart.isdigit() and decipart.isdigit():
                a=int(intpart)+int(decipart)/10**len(decipart)
                return sign*a
    return "This is not a number!"
print(convert("-3.1"))
print(convert(2))
print(convert("-2"))
print(convert("3.ze"))
print(convert("Bonjour"))
            
#Task 1        
def newton_sqrt(a):
    a=convert(a)
    if type(a)!=int and type(a)!=float:
        return convert(a)
    counter=0
    if a<0:
        return "The square root of" +str(a)+ " doesn't exist"
    elif a==0:
        return (0,counter)
    else:
        x=a
        while abs(x*x-a)>=0.000001:
            x=0.5*(x+(a/x))
            counter+=1
        return (round(x,8),counter)
    # return (answer, number_of_steps)

#Task 2
def bisection_sqrt(a):
    a=convert(a)
    if type(a)!=int and type(a)!=float:
        return convert(a)
    counter=0
    if a<0:
        return "The square root doesn't exist"
    elif a==0:
        return (0,counter)
    else:
        low=0
        if a<1:
            high=1
        else:
            high=a
        while high - low >= 0.000001:
            counter=counter+1
            mid = (low + high) / 2
            if mid * mid < a:
                low = mid
            else:
                high = mid
        return (round(mid,6),counter) # return (answer, number_of_steps)


# Task 3: write your code here

#for a=3
print("For a=3:")
print("With newton's method:",newton_sqrt("3"))
print("With bisection's method:",bisection_sqrt(3),"\n")

#for a=5
print("For a=5:")
print("With newton's method:",newton_sqrt(5))
print("With bisection's method:",bisection_sqrt(5))

#for a=100
print("\nFor a=100:")
print("With newton's method:",newton_sqrt(100))
print("With bisection's method:",bisection_sqrt(100))

#For a=0.25
print("\nFor a=0.25:")
print("With newton's method:",newton_sqrt("0.25"))
print("With bisection's method:",bisection_sqrt(0.25))
```

## Extra (optional)

Use Newton's idea for **cube roots**: $x_{\text{new}} = x - \frac{x^3 - a}{3x^2}$. Write `newton_cbrt(a)` and test it with `a = 27`.

```python
def newton_cbrt(a):
    if type(a)!= int and type(a)!= float:
        if a.isdigit():
            a=float(a)
        else:
            return "Enter a valid number"
    counter=0
    if a==0:
        return 0,counter
    else:
        x=a
        while abs(x*x*x-a)>=0.00000001:
            x=x-((x*x*x-a)/(3*(x*x)))
            counter+=1
        return x,counter
print(newton_cbrt("27"))
print(newton_cbrt(27))
```
